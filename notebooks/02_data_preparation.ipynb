```python
import pandas as pd


from src.config import PROCESSED_DATA_DIR, RAW_DATA_PATH


raw_data = pd.read_csv(RAW_DATA_PATH)
```

```python
# copy the raw data to a new dataframe for preprocessing
df = raw_data.copy()
df.head()
```

```python
# remove unnecessary columns
df = df.drop(['Loan_ID'], axis=1)
```

```python
# map the target variable to binary values
df['Loan_Status'] = df['Loan_Status'].map({'Y': 1, 'N': 0})
```

### Feature Engineering

```python
# create a new column 'Total_Income' 
df['Total_Income'] = df['ApplicantIncome'] + df['CoapplicantIncome']
```

```python
df.info()
```

## Save processsed dataset

```python
missing_target_count = int(df['Loan_Status'].isna().sum())
if missing_target_count:
    raise ValueError(
        f"Loan_Status has {missing_target_count} missing value(s); "
        "processed dataset was not saved."
    )

PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)
df.to_csv(PROCESSED_DATA_DIR / "processed_data.csv", index=False)
```
